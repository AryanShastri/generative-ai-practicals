```python
# 01 - Generative AI vs Discriminative Models
# MNIST Classification and Image Reconstruction

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models

# --------------------------------------------------
# 1. Load MNIST Dataset
# --------------------------------------------------

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training data shape:", x_train.shape)
print("Testing data shape:", x_test.shape)

# --------------------------------------------------
# 2. Preprocess the Data
# --------------------------------------------------

# Normalize pixel values from 0-255 to 0-1
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Add channel dimension for CNN
x_train_cnn = np.expand_dims(x_train, -1)
x_test_cnn = np.expand_dims(x_test, -1)

print("Processed training shape:", x_train_cnn.shape)


# ==================================================
# PART A: DISCRIMINATIVE MODEL
# MNIST CLASSIFICATION
# ==================================================

print("\nTraining Discriminative Model...")

# --------------------------------------------------
# 3. Build CNN Classification Model
# --------------------------------------------------

classifier = models.Sequential([
    layers.Input(shape=(28, 28, 1)),

    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(64, activation="relu"),

    # 10 output classes: digits 0-9
    layers.Dense(10, activation="softmax")
])

# --------------------------------------------------
# 4. Compile the Classifier
# --------------------------------------------------

classifier.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

classifier.summary()

# --------------------------------------------------
# 5. Train the Classification Model
# --------------------------------------------------

classifier.fit(
    x_train_cnn,
    y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.1
)

# --------------------------------------------------
# 6. Evaluate the Model
# --------------------------------------------------

test_loss, test_accuracy = classifier.evaluate(
    x_test_cnn,
    y_test,
    verbose=0
)

print("\nClassification Accuracy:",
      round(test_accuracy * 100, 2), "%")


# --------------------------------------------------
# 7. Make Predictions
# --------------------------------------------------

predictions = classifier.predict(x_test_cnn[:10])

predicted_labels = np.argmax(predictions, axis=1)

print("\nActual Labels:   ", y_test[:10])
print("Predicted Labels:", predicted_labels)


# --------------------------------------------------
# 8. Display Classification Results
# --------------------------------------------------

plt.figure(figsize=(12, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)

    plt.imshow(x_test[i], cmap="gray")

    plt.title(
        f"Actual: {y_test[i]}\nPredicted: {predicted_labels[i]}"
    )

    plt.axis("off")

plt.suptitle("MNIST Classification using CNN")
plt.tight_layout()
plt.show()


# ==================================================
# PART B: GENERATIVE MODEL
# AUTOENCODER FOR IMAGE RECONSTRUCTION
# ==================================================

print("\nTraining Generative Model (Autoencoder)...")

# --------------------------------------------------
# 9. Build Encoder
# --------------------------------------------------

encoder = models.Sequential([
    layers.Input(shape=(28, 28, 1)),

    layers.Conv2D(
        32,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    layers.MaxPooling2D(
        (2, 2),
        padding="same"
    ),

    layers.Conv2D(
        16,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    layers.MaxPooling2D(
        (2, 2),
        padding="same"
    )
], name="Encoder")


# --------------------------------------------------
# 10. Build Decoder
# --------------------------------------------------

decoder = models.Sequential([
    layers.Conv2D(
        16,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    layers.UpSampling2D((2, 2)),

    layers.Conv2D(
        32,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    layers.UpSampling2D((2, 2)),

    # Reconstruct original 28x28 grayscale image
    layers.Conv2D(
        1,
        (3, 3),
        activation="sigmoid",
        padding="same"
    )
], name="Decoder")


# --------------------------------------------------
# 11. Create Autoencoder
# --------------------------------------------------

autoencoder = models.Sequential([
    encoder,
    decoder
], name="Autoencoder")

# --------------------------------------------------
# 12. Compile Autoencoder
# --------------------------------------------------

autoencoder.compile(
    optimizer="adam",
    loss="binary_crossentropy"
)

autoencoder.summary()

# --------------------------------------------------
# 13. Train Autoencoder
# --------------------------------------------------

autoencoder.fit(
    x_train_cnn,
    x_train_cnn,
    epochs=5,
    batch_size=128,
    validation_data=(x_test_cnn, x_test_cnn)
)


# --------------------------------------------------
# 14. Reconstruct Images
# --------------------------------------------------

reconstructed_images = autoencoder.predict(
    x_test_cnn[:10]
)


# --------------------------------------------------
# 15. Display Original and Reconstructed Images
# --------------------------------------------------

plt.figure(figsize=(12, 5))

for i in range(10):

    # Original image
    plt.subplot(2, 10, i + 1)

    plt.imshow(
        x_test[i],
        cmap="gray"
    )

    plt.title("Original")
    plt.axis("off")

    # Reconstructed image
    plt.subplot(2, 10, i + 11)

    plt.imshow(
        reconstructed_images[i].squeeze(),
        cmap="gray"
    )

    plt.title("Reconstructed")
    plt.axis("off")

plt.suptitle("MNIST Image Reconstruction using Autoencoder")
plt.tight_layout()
plt.show()


# --------------------------------------------------
# 16. Compare the Two Models
# --------------------------------------------------

print("\n==============================")
print("       MODEL COMPARISON")
print("==============================")

print("\nDiscriminative Model:")
print("Purpose: Classification")
print("Input: MNIST Image")
print("Output: Digit Label (0-9)")

print("\nGenerative Model:")
print("Purpose: Image Reconstruction")
print("Input: MNIST Image")
print("Output: Reconstructed Image")

print("\nBasic Difference:")
print("Discriminative → Decide / Classify")
print("Generative     → Generate / Reconstruct")
```
