```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Set distribution parameters
mu = 50
sigma = 10

# 2. Generate synthetic data via sampling
samples = np.random.normal(mu, sigma, 1000)

# 3. Compute theoretical Gaussian PDF
x_range = np.linspace(10, 90, 1000)
pdf_curve = norm.pdf(x_range, mu, sigma)

# 4. Visualize histogram along with the PDF
plt.figure(figsize=(8, 5))
plt.hist(samples, bins=30, density=True, alpha=0.6, label="Synthetic samples")
plt.plot(x_range, pdf_curve, "r", label="Gaussian PDF")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Gaussian Distribution and Synthetic Data")
plt.legend()
plt.show()
```
